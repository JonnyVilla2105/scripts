```python
buses = pd.read_excel('C:/Users/Jonny Villareal/OneDrive - Gmovil SAS/Escritorio/Jonny/Control 2026/buses_inmovilizados.xlsx')

buses.head(2)
```

```python
# ======================================================
# FECHAS CORRECTAS
# ======================================================

buses_aux = buses.copy()

buses_aux['Fecha_bus'] = pd.to_datetime(
    buses_aux['Fecha de Inmovilización'],
    dayfirst=True,
    errors='coerce'
)

# Dejar SOLO texto dd/mm/yyyy
buses_aux['Fecha_bus'] = buses_aux['Fecha_bus'].dt.strftime('%d/%m/%Y')


df_aux = df_consolidado.copy()

df_aux['Fecha_nov'] = pd.to_datetime(
    df_aux['Fecha Novedad'],
    dayfirst=True,
    errors='coerce'
)

df_aux['Fecha_nov'] = df_aux['Fecha_nov'].dt.strftime('%d/%m/%Y')


# ======================================================
# CREAR COLUMNAS DE FECHAS
# ======================================================

buses_aux['Marca'] = 'X'

tabla_fechas = (
    buses_aux
    .pivot_table(
        index='Placa',
        columns='Fecha_bus',
        values='Marca',
        aggfunc='first'
    )
    .reset_index()
)

# ======================================================
# CRUZAR
# ======================================================

resultado = df_aux.merge(
    tabla_fechas,
    on='Placa',
    how='left'
)

resultado.head()
```
