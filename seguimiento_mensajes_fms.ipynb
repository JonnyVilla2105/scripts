```python
import re
import pandas as pd
import numpy as np
import os
import glob
from pathlib import Path
import shutil
```

```python
dia = '20260522'
```

```python
origen = r"Z:/01 base_datos/41 Informe Diario CCZ/FMS 2025/SMART OPERATOR/mensajes ul-SC smart operator/mensajes SC-ul dia vencido"

# Listar solo archivos CSV del día
archivos_csv = [
    os.path.join(origen, f) 
    for f in os.listdir(origen) 
    if f.endswith(".csv") and dia in f
]

print(f"Archivos encontrados: {len(archivos_csv)}")

# Unir todos en un solo DataFrame
df_final = pd.concat(
    (pd.read_csv(f, sep=';', encoding='latin1',skiprows=1, low_memory=False) for f in archivos_csv),
    ignore_index=True
)
```

```python
ok_files = []
error_files = []

for f in archivos_csv:
    try:
        pd.read_csv(f, sep=';', encoding='latin1', skiprows=1)
        ok_files.append(f)
    except Exception as e:
        error_files.append((f, str(e)))

print("TOTAL OK:", len(ok_files))
print("TOTAL ERROR:", len(error_files))
```

```python
for f, e in error_files:
    print("📁 Archivo con error:")
    print(f)
    print("⚠️ Error:")
    print(e)
```

```python
fin = r"Z:/01 base_datos/41 Informe Diario CCZ/FMS 2025/SMART OPERATOR/mensajes ul-SC smart operator/mensajes SC-ul dia vencido/Consolidado"
```

```python
# Guardar el consolidado en la MISMA CARPETA
archivo_salida = os.path.join(fin, f"{dia}_CONSOLIDADO_ULSC.csv")

df_final.to_csv(archivo_salida, index=False, sep=';')

print(f"✅ Consolidado creado: {archivo_salida}")
```
