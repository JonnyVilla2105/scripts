Asignar la rotación de los turnos

```python
cols_semanas = [col for col in rotacion.columns if 'SEMANA' in col]

rot_dict = {}

for patio, df in rotacion.groupby('PATIO'):
    rot_dict[patio] = df[cols_semanas].values.tolist()
```

```python
def obtener_fila_rotacion(patio, base):
    
    filas = rot_dict.get(patio, [])
    
    for fila in filas:
        if base in fila:
            return fila
    
    return None
```

```python
fila_tipo = personal_prog_final.iloc[1]

for i in range(2, len(personal_prog_final)):
    
    patio = personal_prog_final.loc[i, 'PATIO']
    base = personal_prog_final.loc[i, 'BASE']
    
    if pd.isna(base):
        continue
    
    fila_rot = obtener_fila_rotacion(patio, base)
    
    if fila_rot is None:
        continue
    
    idx_inicio = fila_rot.index(base)
    semana_idx = idx_inicio
    
    primera_asignacion = True
    semana_iso_actual = None
    
    for col in columnas_fechas:
        
        fecha = pd.to_datetime(col, dayfirst=True)
        tipo = str(fila_tipo[col]).lower()
        
        semana_iso = fecha.isocalendar().week
        
        # 🔥 cambio de semana REAL (aunque sea festivo)
        if not primera_asignacion and semana_iso != semana_iso_actual:
            semana_idx = (semana_idx + 1) % len(fila_rot)
        
        semana_iso_actual = semana_iso
        
        # ❌ saltar días no válidos
        if tipo in ['sabado','domingo','festivo']:
            continue
        
        # ❌ no sobreescribir VAC
        if personal_prog_final.loc[i, col] == 'VAC':
            continue
        
        turno = fila_rot[semana_idx]
        
        personal_prog_final.loc[i, col] = turno
        
        primera_asignacion = False

personal_prog_final.head()
```

```python
fila_tipo = personal_prog_final.iloc[1]

for i in range(2, len(personal_prog_final)):
    
    patio = personal_prog_final.loc[i, 'PATIO']
    base = personal_prog_final.loc[i, 'BASE']
    
    if pd.isna(base):
        continue
    
    fila_rot = obtener_fila_rotacion(patio, base)
    
    if fila_rot is None:
        continue
    
    idx_inicio = fila_rot.index(base)
    semana_idx = idx_inicio
    
    primera_asignacion = True
    semana_iso_actual = None
    
    for col in columnas_fechas:
        
        fecha = pd.to_datetime(col, dayfirst=True)
        tipo = str(fila_tipo[col]).lower()
        
        semana_iso = fecha.isocalendar().week
        
        # 🔥 CAMBIO DE SEMANA SIEMPRE (independiente de VAC o tipo)
        if not primera_asignacion and semana_iso != semana_iso_actual:
            semana_idx = (semana_idx + 1) % len(fila_rot)
        
        semana_iso_actual = semana_iso
        
        turno = fila_rot[semana_idx]  # 🔥 SIEMPRE calcular turno
        
        # ❌ NO sobreescribir VAC
        if personal_prog_final.loc[i, col] == 'VAC':
            primera_asignacion = False
            continue
        
        # ❌ no escribir en sábados, domingos, festivos
        if tipo in ['sabado','domingo','festivo']:
            primera_asignacion = False
            continue
        
        # ✅ asignar turno
        personal_prog_final.loc[i, col] = turno
        
        primera_asignacion = False
        
personal_prog_final.head()
```

Rotación asignando turnos faltantes

```python
fila_tipo = personal_prog_final.iloc[1]

turnos_objetivo = [
    'ET1T-T1','ET1N','ET2M-T1','ET2T-T1','ET1M-T1',
    'ET3T-T2','ENLACE N','ET5M-T2','ET5T-T2','ET3M-T2',
    'ET6T-Y','ET6N','ET7M-Y','ET7T-Y','ET6M-Y',
    'ET8T-V','ET3N','ET9M-V','ET9T-V','ET8M-V',
    'ET4I-T2','ET4M-T2','ET4T-T2'
]

# 🔥 FUNCIÓN CORRECTA
def obtener_tipo_turno(turno):
    turno = str(turno).upper().strip()
    
    if '-' in turno:
        base = turno.split('-')[0]  # ejemplo: ET2M
    else:
        base = turno
    
    if base.endswith('M'):
        return 'M'
    elif base.endswith('T'):
        return 'T'
    elif base.endswith('N'):
        return 'N'
    
    return None


for col in columnas_fechas:
    
    tipo_dia = str(fila_tipo[col]).lower()
    
    # ✅ SOLO HÁBIL
    if tipo_dia != 'habil':
        continue

    # valores actuales
    col_vals = personal_prog_final.loc[2:, col].copy()

    # 🔥 candidatos EXACTOS (solo M o T puros)
    candidatos_M = list(col_vals[col_vals == 'M'].index)
    candidatos_T = list(col_vals[col_vals == 'T'].index)

    # turnos actuales
    turnos_actuales = set(
        col_vals.dropna().astype(str).str.strip()
    )

    faltantes = [t for t in turnos_objetivo if t not in turnos_actuales]

    if not faltantes:
        continue

    # 🔥 separar correctamente por tipo REAL
    faltantes_M = [t for t in faltantes if obtener_tipo_turno(t) == 'M']
    faltantes_T = [t for t in faltantes if obtener_tipo_turno(t) == 'T']

    # =========================
    # ✅ ASIGNAR M (solo a M)
    # =========================
    for turno in faltantes_M:
        if not candidatos_M:
            break
        
        i_sel = candidatos_M.pop(0)
        personal_prog_final.loc[i_sel, col] = turno

    # =========================
    # ✅ ASIGNAR T (solo a T)
    # =========================
    for turno in faltantes_T:
        if not candidatos_T:
            break
        
        i_sel = candidatos_T.pop(0)
        personal_prog_final.loc[i_sel, col] = turno

personal_prog_final.head()
```

Visualización de turnos faltantes

```python
fila_tipo = personal_prog_final.iloc[1]

# Diccionario: fecha -> lista de faltantes
pendientes_por_fecha = {}

max_filas = 0

for col in columnas_fechas:
    
    tipo_dia = str(fila_tipo[col]).lower()
    
    if tipo_dia != 'habil':
        pendientes_por_fecha[col] = []
        continue
    
    turnos_actuales = set(
        personal_prog_final.loc[2:, col]
        .dropna()
        .astype(str)
        .str.strip()
    )
    
    faltantes = [t for t in turnos_objetivo if t not in turnos_actuales]
    
    pendientes_por_fecha[col] = faltantes
    
    if len(faltantes) > max_filas:
        max_filas = len(faltantes)



filas_pendientes = []

for n in range(max_filas):
    
    fila = {}
    
    for col in columnas_fechas:
        
        lista = pendientes_por_fecha[col]
        
        if n < len(lista):
            fila[col] = lista[n]
        else:
            fila[col] = ''
    
    filas_pendientes.append(fila)



df_pendientes = pd.DataFrame(filas_pendientes)

# Agregar columnas descriptivas
ids = ['PENDIENTES'] + [f'PENDIENTES {i}' for i in range(1, len(df_pendientes))]

df_pendientes.insert(0, 'Id', ids)
df_pendientes.insert(1, 'Name', '')
df_pendientes.insert(2, 'NOMBRES', '')
df_pendientes.insert(3, 'APELLIDOS', '')
df_pendientes.insert(4, 'CARGO', '')
df_pendientes.insert(5, 'PATIO', '')
df_pendientes.insert(6, 'BASE', '')



personal_prog_final = pd.concat(
    [personal_prog_final, df_pendientes],
    ignore_index=True
)

personal_prog_final.head()
```

Asignar descansos y turnos N de acuerdo al tipo día

```python
fila_tipo = personal_prog_final.iloc[1]

for i in range(2, len(personal_prog_final)):
    
    semana_actual = None
    tiene_nocturno = False
    cols_semana = []
    
    for col in columnas_fechas:
        
        fecha = pd.to_datetime(col, dayfirst=True)
        tipo = str(fila_tipo[col]).lower()
        valor = personal_prog_final.loc[i, col]
        
        semana_iso = fecha.isocalendar().week
        
        # 🔁 cambio de semana
        if semana_actual is not None and semana_iso != semana_actual:
            
            # 🔥 aplicar D si hubo nocturno
            if tiene_nocturno:
                for c in cols_semana:
                    tipo_c = str(fila_tipo[c]).lower()
                    
                    if tipo_c in ['sabado','domingo']:
                        if pd.isna(personal_prog_final.loc[i, c]):
                            personal_prog_final.loc[i, c] = 'D'
            
            # reset
            tiene_nocturno = False
            cols_semana = []
        
        semana_actual = semana_iso
        cols_semana.append(col)
        
        # detectar si hay turno N en la semana
        if isinstance(valor, str) and 'N' in valor:
            tiene_nocturno = True
    
    # 🔥 última semana
    if tiene_nocturno:
        for c in cols_semana:
            tipo_c = str(fila_tipo[c]).lower()
            
            if tipo_c in ['sabado','domingo']:
                if pd.isna(personal_prog_final.loc[i, c]):
                    personal_prog_final.loc[i, c] = 'D'
        
personal_prog_final.head()
```

Completar fin de semana

```python
def completar_por_viernes(personal_prog_final):

    df = personal_prog_final

    columnas_fechas = df.columns[5:]
    fila_tipos = 1
    tipos_dia = df.iloc[fila_tipos][columnas_fechas]

    for idx in df.index[2:]:

        for i, col in enumerate(columnas_fechas):

            # Solo cuando es viernes (Hábil)
            if tipos_dia[col] == 'Habil':

                valor_viernes = df.loc[idx, col]

                if pd.isna(valor_viernes) or valor_viernes == '':
                    continue

                # Buscar en el MISMO dataframe la fila cuyo BASE coincida
                match = df[df['BASE'] == valor_viernes]

                if match.empty:
                    continue

                # Tomar patrón de esa BASE encontrada
                sab_val = match['SABADO'].iloc[0]
                dom_val = match['DOMINGO'].iloc[0]

                # Buscar sábado y domingo de esa semana (hacia adelante)
                sab_col = None
                dom_col = None

                for j in range(i+1, len(columnas_fechas)):
                    col_sig = columnas_fechas[j]
                    tipo_sig = tipos_dia[col_sig]

                    if tipo_sig == 'Sabado':
                        sab_col = col_sig

                    elif tipo_sig == 'Domingo':
                        dom_col = col_sig
                        break

                # Completar SOLO si está vacío
                if sab_col:
                    if pd.isna(df.loc[idx, sab_col]) or df.loc[idx, sab_col] == '':
                        if pd.notna(sab_val) and sab_val != '':
                            df.loc[idx, sab_col] = sab_val

                if dom_col:
                    if pd.isna(df.loc[idx, dom_col]) or df.loc[idx, dom_col] == '':
                        if pd.notna(dom_val) and dom_val != '':
                            df.loc[idx, dom_col] = dom_val

    return df


# Aplicar directamente
personal_prog_final = completar_por_viernes(personal_prog_final)

personal_prog_final.head()
```

```python
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

# Cargar archivo
wb = load_workbook(ruta_archivo)
ws = wb.active

# 🎨 Colores
color_vac = PatternFill(start_color="ADD8E6", end_color="ADD8E6", fill_type="solid")   # Azul
color_d = PatternFill(start_color="C0C0C0", end_color="C0C0C0", fill_type="solid")     # Gris
color_habil = PatternFill(start_color="90EE90", end_color="90EE90", fill_type="solid") # Verde
color_no_habil = PatternFill(start_color="FFF59D", end_color="FFF59D", fill_type="solid") # Amarillo

for row in ws.iter_rows(min_row=2):  # saltar encabezado
    
    for cell in row:
        
        if not isinstance(cell.value, str):
            continue
        
        valor = cell.value.strip()
        
        # 🔵 VAC
        if valor == "VAC":
            cell.fill = color_vac
        
        # ⚪ D
        elif valor == "D":
            cell.fill = color_d
        
        # 🟢 HABIL
        elif valor.lower() == "habil":
            cell.fill = color_habil
        
        # 🟡 NO HÁBIL
        elif valor.lower() in ["festivo", "sabado", "domingo"]:
            cell.fill = color_no_habil

# Guardar cambios
wb.save(ruta_archivo)
```
