```python
import pandas as pd
import numpy as np
import subprocess
import os
from datetime import datetime, timedelta, time, date
import datetime
from datetime import datetime
import locale
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Datos de filtros
fecha_ip = '20260501'
fecha_fp = '20260531'

mes = 'may'
```

```python
# Ruta de la carpeta que contiene los archivos
ruta_carpeta = 'Z:/01 base_datos/18 pasajeros_asociados'

# Fechas de inicio y fin para el filtro
fecha_inicio = f'{fecha_ip}'
fecha_fin = f'{fecha_fp}'

# Lista para almacenar los DataFrames
dataframes = []

# Recorrer todos los archivos en la carpeta
for nombre_archivo in os.listdir(ruta_carpeta):
    if nombre_archivo.endswith('_pax_asociados.csv'):
        # Extraer la fecha del nombre del archivo
        fecha_archivo = nombre_archivo[:8]  
        
        # Convertir la fecha a un formato adecuado para comparación
        fecha_archivo_dt = pd.to_datetime(fecha_archivo, format='%Y%m%d', errors='coerce')

        # Verificar si la fecha está dentro del rango deseado
        if fecha_inicio <= fecha_archivo <= fecha_fin:
            ruta_archivo = os.path.join(ruta_carpeta, nombre_archivo)
            df = pd.read_csv(ruta_archivo, encoding='latin', low_memory=False)
            dataframes.append(df)

# Verificar si se encontraron DataFrames
if dataframes:
    # Consolidar todos los DataFrames en uno solo
    resultado2 = pd.concat(dataframes, ignore_index=True)
else:
    print("No se encontraron archivos para consolidar.")
    
resultado2.head()
```

```python
# Dividir la columna "FechaReal" en dos columnas "Fecha" y "Hora"
resultado2[['Fecha_real', 'Hora']] = resultado2['FechaReal'].str.split('T', expand=True)

# Eliminar el carácter "Z" al final de la columna "Hora"
resultado2['Hora'] = resultado2['Hora'].str.rstrip('Z')

# Dividir la columna 'hora' en partes usando ':', y seleccionar la primera parte (horas)
resultado2['franja'] = resultado2['Hora'].str.split(':').str[0]

resultado2.head()
```

```python
# Eliminar las columnas
columnas_a_eliminar = ['FechaId','IdViajeGrupo','HoraInicio','TipoOperacion']
resultado2 = resultado2.drop(columns=columnas_a_eliminar)

resultado2.head()
```

```python
# Convertir la columna 'Fecha' a tipo datetime
resultado2['Fecha_real'] = pd.to_datetime(resultado2['Fecha_real'], format='%Y-%m-%d')

# Extraer el nombre del mes en español
resultado2['Mes'] = resultado2['Fecha_real'].dt.strftime('%B').str.capitalize()

# Extraer el año y almacenarlo como un entero en una nueva columna 'Año'
resultado2['Año'] = resultado2['Fecha_real'].dt.year

# Formatear la columna 'Fecha' en el formato dd/mm/yyyy
resultado2['Fecha_real'] = resultado2['Fecha_real'].dt.strftime('%d/%m/%Y')

# Mostrar el DataFrame actualizado
resultado2.head()
```

```python
# Limpiar la columna 'Hora' eliminando caracteres no deseados
resultado2['Hora'] = resultado2['Hora'].replace('[^\d:]', '', regex=True)

# Reemplazar el caso del punto con un valor predeterminado (por ejemplo, '0:00:00')
resultado2['Hora'] = resultado2['Hora'].replace('.', '0:00:00')

# Convertir la columna 'Hora Petición' a timedelta y luego a segundos
resultado2['HoraIniSegundos'] = pd.to_timedelta(resultado2['Hora']).dt.total_seconds()

# Asignar un valor constante a la nueva columna 'cantidad'
resultado2['cantidad'] = 1 

resultado2.head()
```

```python
cols = ['LineAso','RutaAso','RutaOrg','IdViaje','Conductor','HoraIniSegundos']

resultado2[cols] = (
    resultado2[cols]
    .fillna(0)
    .astype(int)
)
resultado2.head()
```

```python
resultado_resumido = (
    resultado2
    .groupby([
        'Fecha','NumeroBus','ServicioBus','LineAso',
        'RutaAso','RutaOrg','LineaOrg','IdViaje','Conductor'
    ], as_index=False)
    ['cantidad']
    .sum()
)
```

```python
resultado_resumido = resultado_resumido.rename(
    columns={'cantidad': 'validaciones'}
)

resultado_resumido.head()
```
