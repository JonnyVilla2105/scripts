```python
import pandas as pd
import numpy as np
```

```python
matriz = pd.read_csv('Z:/01 base_datos/06 matriz_distancia_FMS/20260529_matriz distancias.csv', encoding='latin')

matriz
```

```python
# function to find the value closest to the 25th percentile
def closest_to_25_percentile(group):
    q25 = np.percentile(group['PosiciÃ³n'], 25)
    closest_value = group.iloc[(group['PosiciÃ³n'] - q25).abs().argsort().iloc[0]]
    return closest_value

#Apply for group
result = matriz.groupby(['Id LÃ­nea', 'Id Ruta'], group_keys=False).apply(closest_to_25_percentile)

result
```
