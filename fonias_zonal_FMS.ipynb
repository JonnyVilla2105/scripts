```python
def fix_time(date_str):
    # Strip any leading or trailing whitespace
    date_str = date_str.strip()
    
    # Ensure the string has the correct format
    if ':' not in date_str or len(date_str.split(':')) < 3:
        return date_str
    
    hour = date_str.split(':')[0]
    
    # Check if the hour part is numeric and convert it
    if hour.isdigit() and int(hour) >= 24:
        corrected_hour = int(hour) - 24
        corrected_time = f'{corrected_hour:02}' + date_str[2:]
        return corrected_time
    return date_str

ab['Hora'] = ab['Hora'].astype(str)

ab['Hora'] = ab['Hora'].apply(fix_time)

ab['Hora'] = pd.to_datetime(
    ab['Hora'],
    format='%H:%M:%S',
    errors='coerce'
).dt.time

ab.head()
```

```python
# Asegúrate de que las columnas 'Hora Apertura' y 'Hora Petición' están en formato de tiempo
fonias['Tiempo Apertura'] = pd.to_datetime(fonias['Tiempo Apertura'], errors='coerce').dt.time
fonias['Tiempo Petición Real'] = pd.to_datetime(fonias['Tiempo Petición Real'], errors='coerce').dt.time

# Función para calcular el tiempo de atención y formatearlo como 'HH:MM:SS'
def calcular_tiempo_atencion(row):
    if pd.isna(row['Tiempo Apertura']) or row['Estado'] == 'Eliminado de la lista':
        return ''
    
    hora_apertura = pd.to_datetime(row['Tiempo Apertura'], format='%H:%M:%S').time()
    hora_peticion = pd.to_datetime(row['Tiempo Petición Real'], format='%H:%M:%S').time()
    
    # Convertir las horas a datetime para la resta
    t1 = pd.to_datetime(hora_apertura.strftime('%H:%M:%S'), format='%H:%M:%S')
    t2 = pd.to_datetime(hora_peticion.strftime('%H:%M:%S'), format='%H:%M:%S')

    if hora_apertura == hora_peticion:
        return '00:00:00'
    elif hora_peticion < hora_apertura:
        diferencia = t1 - t2
    elif hora_peticion < pd.to_datetime('23:59:00', format='%H:%M:%S').time() and hora_apertura > pd.to_datetime('00:00:00', format='%H:%M:%S').time():
        t1 = pd.to_datetime('23:59:00', format='%H:%M:%S')
        t2 = pd.to_datetime(hora_peticion.strftime('%H:%M:%S'), format='%H:%M:%S')
        t3 = pd.to_datetime(hora_apertura.strftime('%H:%M:%S'), format='%H:%M:%S')
        diferencia = (t1 - t2) + (t3 - pd.to_datetime('00:00:00', format='%H:%M:%S'))
    else:
        return ''

    # Formatear la diferencia como 'HH:MM:SS'
    return str(diferencia).split()[-1]

# Aplicar la función para crear la nueva columna
fonias['TMR'] = fonias.apply(calcular_tiempo_atencion, axis=1)

# Mostrar el DataFrame resultante
fonias.head()
```

```python
# Función para convertir un tiempo en formato datetime a segundos desde el inicio del día
def time_to_seconds(time):
    if pd.isna(time):
        return None
    return time.hour * 3600 + time.minute * 60 + time.second

# Aplicar la función a las columnas correspondientes
fonias['Hora Petición Segundos'] = fonias['Tiempo Petición Real'].apply(time_to_seconds)
acb['HoraMinima Segundos'] = acb['HoraMinima'].apply(time_to_seconds)
acb['HoraMaxima Segundos'] = acb['HoraMaxima'].apply(time_to_seconds)
```

```python
def calcular_max_linea(fecha, codigo_bus, hora_peticion_segundos):
    # Filtro para el DataFrame combinado
    filtro = (
        (acb['Fecha'] == fecha) &
        (acb['Código Bus'] == codigo_bus) &
        (acb['HoraMinima Segundos'] <= hora_peticion_segundos) &
        (hora_peticion_segundos < acb['HoraMaxima Segundos'])
    )
    
    # Filtrar el DataFrame acb según el filtro
    resultado = acb.loc[filtro, 'Viaje Linea']
    
    # Verificar si hay algún resultado después de aplicar el filtro
    if not resultado.empty:
        # Obtener el valor máximo de 'ViajeLinea'
        return resultado.max()
    
    return None  # Devolver None si no hay resultados

# Aplica la función a las columnas correspondientes en fonias
fonias['Viaje'] = fonias.apply(
    lambda row: calcular_max_linea(
        row['Fecha'],
        row['Código Bus'],
        row['Hora Petición Segundos']
    ),
    axis=1
)

fonias['Viaje'].fillna(0, inplace=True)

# Convertir la columna 'Viaje' a enteros
fonias['Viaje'] = fonias['Viaje'].astype(int)

fonias.head()
```
